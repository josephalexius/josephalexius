# Data Cleaning and Feature Engineering

```python
#group by YY1 column then aggregate all columns.
#mode applied to categorical, mean to numerical
working_dataset = df[['YY1','AGECL','EDCL','MARRIED','KIDS','KNOWL','OCCAT1','FAMSTRUCT','HHSEX','NOFINRISK','YESFINRISK', 'ASSET','INCOME','NETWORTH','EQUITY','DEBT']]
working_dataset = df.groupby('YY1').agg(
    AGECL = ('AGECL', lambda x: x.mode().iloc[0]),
    EDCL = ('EDCL', lambda x: x.mode().iloc[0]),
    MARRIED = ('MARRIED', lambda x: x.mode().iloc[0]),
    KIDS = ('KIDS', lambda x: x.mode().iloc[0]),
    KNOWL = ('KNOWL', lambda x: x.mode().iloc[0]),
    OCCAT1 = ('OCCAT1', lambda x: x.mode().iloc[0]),
    FAMSTRUCT = ('FAMSTRUCT', lambda x: x.mode().iloc[0]),
    HHSEX = ('HHSEX', lambda x: x.mode().iloc[0]),
    NOFINRISK = ('NOFINRISK', lambda x: x.mode().iloc[0]),
    YESFINRISK = ('YESFINRISK', lambda x: x.mode().iloc[0]),
    ASSET = ('ASSET','mean'),
    INCOME = ('INCOME','mean'),
    NETWORTH = ('NETWORTH','mean'),
    EQUITY = ('EQUITY','mean'),
    DEBT = ('DEBT','mean')
).reset_index()
```

```python
working_dataset.head(20)
```

```python
working_dataset['KNOWL'] = pd.to_numeric(working_dataset['KNOWL'])
```

```python
#bin KNOWL into 3 categories
# -1, 0-3 (Little to no knowldedge in investing, 1)
# 4-7 (Medium knowledge in investing, 2)
# 8-10 (Advanced knowledge in investing, 3)
working_dataset['Knowledge Group'] = np.where(
    working_dataset['KNOWL'] <=3, 1, np.where(
        (working_dataset['KNOWL'] >3) & (working_dataset['KNOWL'] <=7), 2, 3 
    )
)

#derive metrics from dataset
#compare networth to asset 
working_dataset['Equity to Asset Ratio'] = working_dataset['EQUITY'] / (working_dataset['ASSET'] + 1)
#compare debt to asset
working_dataset['Debt to Asset Ratio'] = working_dataset['DEBT'] / (working_dataset['ASSET'] + 1)
#compare debt to income
working_dataset['Debt to Income Ratio'] = working_dataset['DEBT'] / (working_dataset['INCOME'] + 1)


##### metrics scoring section #####

#translate metrics into categories that is aligned with vanguard risk profile to prepare for classification
#equity ratio
# if ratio is > 0.70, score 2
# if ratio is < 0.70 and > 0.30, score 1
# if ratio is < 0.30, score 0
working_dataset['Equity to Asset Ratio Score'] = np.where(
    working_dataset['Equity to Asset Ratio']>0.70, 2, (np.where(
        (working_dataset['Equity to Asset Ratio']<0.70) & (working_dataset['Equity to Asset Ratio']>0.30),1,0)))

#debt to asset ratio
# if ratio < 0.10, score 2
# if ratio is > 0.10 and < 0.40, score 1
# if ratio is > 0.40, score 0
working_dataset['Debt to Asset Ratio Score'] = np.where(
    working_dataset['Debt to Asset Ratio']<0.10, 2, (np.where(
        (working_dataset['Debt to Asset Ratio']>0.10) & (working_dataset['Equity to Asset Ratio']<0.40),1,0)))

#debt to income ratio
# if ratio < 0.20, score 2
# if ratio is > 0.20 and < 0.40, score 1
# if ratio is > 0.40, score 0
working_dataset['Debt to Income Ratio Score'] = np.where(
    working_dataset['Debt to Income Ratio']<0.20, 2, (np.where(
        (working_dataset['Debt to Income Ratio']>0.20) & (working_dataset['Equity to Asset Ratio']<0.40),1,0)))

#classify risk profile
# >60% treshold source: https://www.investopedia.com/ask/answers/021215/what-good-debt-ratio-and-what-bad-debt-ratio.asp#:~:text=There%20is%20no%20one%20figure,Company%20Improve%20Its%20Debt%20Ratio?
# 0 = Conservative
# 1 = Moderate
# 2 = Aggressive
working_dataset['Risk Profile'] = np.where(
    (working_dataset['Debt to Asset Ratio'] > 0.8) & (working_dataset['Debt to Income Ratio'] > 0.6), 'Conservative', np.where(
        working_dataset['Equity to Asset Ratio Score'] + working_dataset['Debt to Asset Ratio Score'] + working_dataset['Debt to Income Ratio Score'] < 3, 'Conservative', np.where(
            (working_dataset['Equity to Asset Ratio Score'] + working_dataset['Debt to Asset Ratio Score'] + working_dataset['Debt to Income Ratio Score'] >= 3) &
            (working_dataset['Equity to Asset Ratio Score'] + working_dataset['Debt to Asset Ratio Score'] + working_dataset['Debt to Income Ratio Score'] <= 5), 'Moderate', 'Aggressive'
            )
        )
    )

working_dataset['Risk Profile N'] = np.where(
    (working_dataset['Debt to Asset Ratio'] > 0.8) & (working_dataset['Debt to Income Ratio'] > 0.6), '0', np.where(
        working_dataset['Equity to Asset Ratio Score'] + working_dataset['Debt to Asset Ratio Score'] + working_dataset['Debt to Income Ratio Score'] < 3, '0', np.where(
            (working_dataset['Equity to Asset Ratio Score'] + working_dataset['Debt to Asset Ratio Score'] + working_dataset['Debt to Income Ratio Score'] >= 3) &
            (working_dataset['Equity to Asset Ratio Score'] + working_dataset['Debt to Asset Ratio Score'] + working_dataset['Debt to Income Ratio Score'] <= 5), '1', '2'
            )
        )
    )
```

# Model Selection

## 1. Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
```

```python

#instantiate model
model = RandomForestClassifier(class_weight='balanced', random_state=42)

#fit
model.fit(x_train, y_train)

#run predictions
y_pred = model.predict(x_test)

#run test report
print(classification_report(y_test, y_pred))
```

Next step: Apply and test another model, XGBoost.

### Choosing another model: XGBoost

I chose XGBoost as the next model because it can fix errors by sequentially building trees and try to correct the errors from the previous builds instead of running the trees in parallel. Based from mcpanalytics, XGBoost adds regularization terms (L1 and L2 penalties on leaf weights, max depth, minimum child weight) to prevent individual trees from fitting noise which can optimze the results.

```python
# include asset, income, networth, equity, debt instead
xgb_final_dataset = working_dataset[['AGECL','EDCL','MARRIED','KIDS','Knowledge Group','OCCAT1','FAMSTRUCT','HHSEX', 'ASSET', 'INCOME', 'NETWORTH', 'EQUITY', 'DEBT', 'Risk Profile N']]
xgb_final_dataset = working_dataset[['AGECL','EDCL','MARRIED','KIDS','Knowledge Group','OCCAT1','FAMSTRUCT','HHSEX', 'ASSET', 'INCOME', 'NETWORTH', 'EQUITY', 'DEBT', 'Risk Profile N']].astype(int)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.utils.class_weight import compute_sample_weight

#segregate independent and dependent variables
x_xgb = xgb_final_dataset.drop('Risk Profile N', axis=1).copy()
y_xgb = xgb_final_dataset['Risk Profile N']

#use log to convert outliers into a normal distribution
finance_cols = ['ASSET', 'INCOME', 'NETWORTH', 'EQUITY', 'DEBT']
for col in finance_cols:
    x_xgb[col] = np.log1p(x_xgb[col].clip(lower=0))

#perform train test split, then stratify for a proportioned train and test set
x_xgb_train, x_xgb_test, y_xgb_train, y_xgb_test = train_test_split(x_xgb, y_xgb, test_size=0.2, random_state=42, stratify=y_xgb)

#apply normalization to bring everything to a mean of zero and standard deviation of 1.
scaler = StandardScaler()
x_xgb_train[finance_cols] = scaler.fit_transform(x_xgb_train[finance_cols])
x_xgb_test[finance_cols] = scaler.transform(x_xgb_test[finance_cols])
```

```python
y_xgb_train.value_counts()
```

```python
y_xgb_test.value_counts()
```

```python
# Convert attributes to integer
y_xgb_train = y_xgb_train.astype(int)
y_xgb_test = y_xgb_test.astype(int)

# auto calculate weights for xgboost. automatically assign weights. weights are indicators of penalty to increase 
# the models sensitivity to rare data (lesser population e.g. aggressive risk appetite).
weights = compute_sample_weight(class_weight='balanced', y=y_xgb_train)

# Initialize XGBoost, objective set to a multi classified problem
from xgboost import XGBClassifier
model_xgb = XGBClassifier(
    n_estimators=100,
    learning_rate=0.05,
    max_depth=6,
    objective='multi:softprob',
    random_state=42
)

#fit and apply calculated weights
model_xgb.fit(x_xgb_train, y_xgb_train, sample_weight=weights)

#predict
y_xgb_pred = model_xgb.predict(x_xgb_test)

#run test report
print(classification_report(y_xgb_test, y_xgb_pred))
```

Findings: The model overall has a 94% score. However, the precision score for agressive risk profile is only 0.39 or 40%. The F1 Score also is low to 51%.
<br>Target: Raise precision to higher percentage.

### Model Optimization

```python
#try to assign manual weights and reapply the model
manual_weights = y_xgb_train.map({0: 1.0, 1: 1.0, 2: 5.0})

# Re-train with the manual weights
model_xgb.fit(x_xgb_train, y_xgb_train, sample_weight=manual_weights)

y_xgb_pred = model_xgb.predict(x_xgb_test)
print(classification_report(y_xgb_test, y_xgb_pred))
```

```python
# Create a DataFrame with the actual values and the predictions
results_df = pd.DataFrame({
    'Actual Profile': y_xgb_test.values,
    'Predicted Profile': y_xgb_pred
})

# Add a 'Correct' column for easy scanning
results_df['Match'] = results_df['Actual Profile'] == results_df['Predicted Profile']

print(results_df.head(10))
```

    - Missing data in the survey have been imputed five times using a multiple imputation technique.

Seven observations were deleted for the public version of the data set for purposes of disclosure avoidance; thus, there are 22,975 records in the public data set for 4,595 families.
